### Data Cleaning:-

```python
import numpy as np
import pandas as pd
```

```python
df1=pd.read_csv('OriginalData.csv')
df1.head()
```

```python
df1.shape
```

```python
df1.drop(['Balcony','Unnamed: 0','latitude','longitude','Status','neworold','Landmarks','desc'],axis='columns',inplace=True)
```

```python
df1.drop(['Lift','parking','area'],axis='columns',inplace=True)
```

```python
df1.drop(['Furnished_status'],axis='columns',inplace=True)
```

```python
df1.drop(['Bathrooms'],axis='columns',inplace=True)
df1
```

```python
df1.head()
```

```python
df1.isnull().sum()
```

```python
print(df1['Bedrooms'].unique())
print(df1['Address'].unique())
print(df1['type_of_building'].unique())
```

```python
df1.groupby('Address')['Address'].agg('count').sort_values(ascending=False).head(20)
```

```python
df1.Address=df1.Address.apply(lambda x: x.strip())
location_stats=df1.groupby('Address')['Address'].agg('count').sort_values(ascending=False)
```

```python
locationStatsLessThan10=location_stats[location_stats<=10].count()
```

```python
len(df1.Address.unique())
```

```python
# 1. Get value counts of each location
location_counts = df1['Address'].value_counts()

# 2. Identify locations that appear less than 10 times
locations_to_replace = location_counts[location_counts <= 15].index

# 3. Replace those locations with 'other'
df1['Address'] = df1['Address'].apply(lambda x: 'other' if x in locations_to_replace else x)
```

```python
df1.groupby('Address')['Address'].agg('count').sort_values(ascending=False)
```

```python
import matplotlib.pyplot as plt
plt.hist(df1.Price_sqft)
plt.xlabel('Price per Sq. feet')
plt.ylabel('Count')
```

```python
df1['area']=df1['price']/df1['Price_sqft']
df1.drop(['Price_sqft'],axis='columns',inplace=True)
```

```python
df1.rename(columns={'Bedrooms': 'BHK'}, inplace=True)
```

```python
df1.rename(columns={'Address': 'location'}, inplace=True)
df1.head()
```

```python
location_dummies=pd.get_dummies(df1.location).astype(int)
location_dummies.head()
buildType_dummies=pd.get_dummies(df1.type_of_building).astype(int)
buildType_dummies.head()
```

```python
df1=pd.concat([df1,buildType_dummies.drop('Flat',axis='columns')],axis='columns')
df1
```

```python
df1=pd.concat([df1,location_dummies.drop('other',axis='columns')],axis='columns')
df1.head(2)
```

```python
df=df1.drop(['location','type_of_building'],axis='columns')
```

```python
df.head()
```

```python
X=df.drop(['price'],axis='columns')
X.head(2)
```

```python
Y=df.price
Y.head(2)
```

```python
df.to_csv('final_data.csv', index=False)   # Expoting CSV for future use
```

### Model Training:-      (Day 2)

```python
from sklearn.model_selection import train_test_split
```

```python
import sklearn
```

```python
df=pd.read_csv('final_data.csv')
df.head()
```

```python
X=df.drop(['price'],axis='columns')
Y=df.price
```

```python
X_train, X_test, Y_train, Y_test= train_test_split(X,Y, test_size=0.2, random_state=10)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model=LinearRegression()
model.fit(X_train, Y_train)
model.score(X_test, Y_test)
```

```python
model.score(X_train,Y_train)
```

```python
from sklearn.model_selection import ShuffleSplit
from sklearn.model_selection import cross_val_score
```

```python
cv= ShuffleSplit(n_splits=10, random_state=0, test_size=0.2)
cross_val_score(LinearRegression(), X, Y, cv=cv)
```

```python
# taking out the best split
splits = list(cv.split(X, Y))
train_index, test_index = splits[5]
X_train, X_test = X.iloc[train_index], X.iloc[test_index]
Y_train, Y_test = Y.iloc[train_index], Y.iloc[test_index]
model2 = LinearRegression()
model2.fit(X_train, Y_train)
model2.score(X_test, Y_test)
```

```python
def predict_price(BHK, area, individualHouseOrNot1or0, location):
    loc_index= np.where(X.columns==location)[0][0]
    x=np.zeros(len(X.columns))
    x[0]=BHK
    x[1]=area
    x[2]=individualHouseOrNot1or0
    if loc_index>=0:
        x[loc_index]=1
    return (model2.predict([x])[0])/100000
```

```python
print('Predictes Price (in lakhs): ',predict_price(3, 1500, 0, 'Crossings Republik, Ghaziabad, Delhi NCR'))
```
